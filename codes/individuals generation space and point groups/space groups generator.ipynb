```python
import gemmi
from rdflib import RDFS, RDF, Literal, OWL, XSD, Graph, Namespace
```

## Getting point groups

```python

# Gather unique point groups by inspecting all 230 space groups
point_groups = set()

for i in range(1, 231):
    sg = gemmi.SpaceGroup(i)
    # Get the Hermann-Mauguin point group symbol
    hm_symbol = sg.point_group_hm()
    point_groups.add(hm_symbol)

# Sort and print all 32 crystallographic point groups
sorted_pgs = sorted(list(point_groups))

print(f"Total Point Groups Found: {len(sorted_pgs)}\n")
for i, pg in enumerate(sorted_pgs):
    print(i, pg)
```

```python
def add_point_group(hm_symbol, num: int = 407): 

	point_group_URI = CRYO[f"CRYO_0000{str(num)}"]
	g.add((point_group_URI, RDF.type, OWL.NamedIndividual))
	g.add((point_group_URI, RDF.type, CRYO['CRYO_0000406'])) #point group type
	
	lit = f"{hm_symbol} point group"
	g.add((point_group_URI, RDFS.label, Literal(lit, datatype=XSD.string)))
	g.add((point_group_URI, CRYO['CRYO_0000146'], Literal(hm_symbol, datatype=XSD.string)))# has hm symbol
	return point_group_URI
```

```python
uri_mapping = {}
seen_point_groups = set()

```

```python
CRYO = Namespace("https://w3id.org/pmd/cryo/")
g = Graph()
g.bind("cryo", CRYO)
```

```python
failed = []
num = 407
for pg in sorted_pgs:
	if pg in seen_point_groups:
		continue
	try: 
		uri = add_point_group(hm_symbol=pg, num=num)
		uri_mapping[pg] = str(uri)
		seen_point_groups.add(pg)
	except: 
		failed.append(pg)
	num += 1
	
```

## Adding space groups

```python
import gemmi

pg_mapping = {}

for i in range(1, 231):
    sg = gemmi.SpaceGroup(i)
    hm_pg = sg.point_group_hm()
    
    # Access Schoenflies symbol from CCP4 entry metadata
    entry = sg.ccp4_entry()
    if entry:
        sf_sg = entry.schoenflies  # e.g., 'C1^1', 'O^5_h'
        pg_mapping[hm_pg] = sf_sg

for hm, sf in sorted(pg_mapping.items()):
    print(f"HM: {hm:<10} | Space Group Schoenflies Ref: {sf}")
```

```python
uri_mapping = {}
seen_space_groups = set()
```

```python
CRYO = Namespace("https://w3id.org/pmd/cryo/")
num = 149
```

```python
g = Graph()
g.bind("cryo", CRYO)
```

```python
def add_space_group(entry, num: int = 149): 

	space_group_URI = CRYO[f"CRYO_0000{str(num)}"]
	g.add((space_group_URI, RDF.type, OWL.NamedIndividual))
	g.add((space_group_URI, RDF.type, CRYO['CRYO_0000142']))
	
	lit = f"{entry.hall} space group"
	g.add((space_group_URI, RDFS.label, Literal(lit, datatype=XSD.string)))
	g.add((space_group_URI, CRYO['CRYO_0000145'], Literal(entry.number, datatype=XSD.integer)))
	if entry.hm:
		g.add((space_group_URI, CRYO['CRYO_0000146'], Literal(entry.hm, datatype=XSD.string)))
	else: 
		print(f"{entry.number} doesn't have hm")
	if entry.hall:
		g.add((space_group_URI, CRYO['CRYO_0000147'], Literal(entry.hall, datatype=XSD.string)))
	else: 
		print(f"{entry.number} doesn't have hall")
	return space_group_URI
```

```python
failed = []
for entry in gemmi.spacegroup_table_itb():
	if entry.number in seen_space_groups:
		continue
	try: 
		uri = add_space_group(entry, num = num)
		uri_mapping[entry.hall] = str(uri)
		seen_space_groups.add(entry.number)
	except: 
		failed.append(entry)
	num += 1
	
```
